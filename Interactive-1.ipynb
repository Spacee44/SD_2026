```python
lista = ["a","b","c","d","e","f","g"] 
#Hay que enumerar una lista Metodo enumerate

#Algo
for indice, letra in enumerate(lista): 
    print(f"{indice}: {letra}") #Iterar una lista no un indice

n=0; 
while n < len(lista):
     print(f"{n}: {lista[n]}")
     n += 1
```
